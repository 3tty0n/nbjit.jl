# nbjit.jl Demo - Split JIT Compilation for Jupyter Notebooks

This notebook demonstrates the IJulia integration for nbjit.jl, a split-compilation system that allows selective recompilation of code sections marked with `@hole`.

## Setup

First, let's load the IJulia integration module:

```julia
include("../src/ijulia_integration.jl")
using .IJuliaIntegration
```

## Example 1: Basic Usage with Auto-Detected Cell IDs

The `@jit` macro automatically detects the cell ID from IJulia's execution context. You can mark sections of code with `@hole` to enable selective recompilation.

```julia
@jit begin
    base = 42
    @hole delta = 1
    value = base + delta
end
```

When you run the above cell for the first time, both the main block and the hole are compiled. The output shows:
- `main` is **recompiled** (first time)
- `hole 1` is **recompiled** (first time)

Now, let's run the same cell again without changes:

```julia
@jit begin
    base = 42
    @hole delta = 1
    value = base + delta
end
```

This time, both the main block and hole are **cached** - no recompilation needed!

Now let's change only the hole content:

```julia
@jit begin
    base = 42
    @hole delta = 5
    value = base + delta
end
```

Notice that:
- `main` is **cached** (unchanged)
- `hole 1` is **recompiled** (changed)

Only the hole was recompiled, saving compilation time!

## Example 2: Multiple Holes

You can use multiple `@hole` markers in a single code block:

```julia
@jit begin
    start = 5
    @hole step = start + 1
    @hole total = step * 2
    answer = start + total
end
```

Now let's modify only the first hole:

```julia
@jit begin
    start = 5
    @hole step = start + 10
    @hole total = step * 2
    answer = start + total
end
```

You'll see:
- `main` is **cached**
- `hole 1` is **recompiled** (changed)
- `hole 2` is **cached** (unchanged)

Only the modified hole is recompiled!

## Example 3: Using Manual Cell IDs

If you prefer explicit control over cell IDs, use the `@ijit` macro:

```julia
@ijit :my_cell begin
    x = 10
    @hole y = 2
    z = x + y
end
```

## Example 4: Triggering Full Recompilation

Changing the main block structure triggers a full recompilation:

```julia
@jit begin
    base = 42
    @hole delta = 5
    value = base + delta
end
```

Now let's add a new line in the main block:

```julia
@jit begin
    base = 42
    temp = base + 1
    @hole delta = temp
    value = base + delta
end
```

Both the main block and all holes are recompiled because the main structure changed.

## Inspecting the Session

You can inspect the current notebook session to see what's cached:

```julia
session = IJuliaIntegration.current_session()
println("Cached cells: ", keys(session.cells))
```

## Summary

The nbjit.jl IJulia integration provides:

- **Automatic cell tracking** with the `@jit` macro
- **Selective recompilation** - only changed `@hole` sections are recompiled
- **Efficient notebook workflow** - fast iteration on specific code sections
- **Transparent caching** - automatic detection of when to reuse vs. recompile

This is particularly useful for:
- Experimenting with algorithm parameters in hot loops
- Iterative data analysis workflows
- Large notebooks where full recompilation is expensive
